**7.3.** Consider: |0⟩ ── H ── X ── H ── Measure. Before running the circuit:
a) Predict the final state.
b) Predict the measurement result.

c) Explain your reasoning.

Then implement and verify your prediction using Qiskit.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
import numpy as np


qc = QuantumCircuit(1, 1)

qc.h(0) # First Hadamard gate
qc.x(0) # Pauli-X gate
qc.h(0) # Second Hadamard gate

state = Statevector.from_instruction(qc)
print("--- Verification (Theoretical State) ---")
print(f"Final Statevector (Amplitudes):\n{state}")
print(f"Probabilities from Statevector: {state.probabilities()}\n")
qc.measure(0, 0)
simulator = AerSimulator()
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts(qc)

print("--- Verification (Experimental Measurement) ---")
print(f"Measurement results (1000 shots): {counts}")
```

**7.4.** Create a circuit that puts one qubit into superposition. Obtain:
a) The statevector.
b) The theoretical probabilities.

c) The measurement counts for 1000 shots.

Prepare a table:


```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
import numpy as np

qc_state = QuantumCircuit(1)

statevector = Statevector.from_instruction(qc_state)

theoretical_probabilities = statevector.probabilities()
theoretical_p0 = theoretical_probabilities[0]
theoretical_p1 = theoretical_probabilities[1]


qc_measure = QuantumCircuit(1, 1)
qc_measure.h(0) # Apply Hadamard gate for superposition
qc_measure.measure(0, 0) # Measure the qubit

simulator = AerSimulator()
shots = 1000
result = simulator.run(qc_measure, shots=shots).result()
counts = result.get_counts(qc_measure)

experimental_p0 = counts.get('0', 0) / shots
experimental_p1 = counts.get('1', 0) / shots

# --- Prepare a table ---

print(f"{"Quantity":<20} {"Result":<30}")
print("-" * 50)
print(f"{"Statevector":<20} {str(statevector):<30}")
print(f"{"Theoretical P(0)":<20} {theoretical_p0:<30.3f}")
print(f"{"Theoretical P(1)":<20} {theoretical_p1:<30.3f}")
print(f"{"Experimental P(0)":<20} {experimental_p0:<30.3f}")
print(f"{"Experimental P(1)":<20} {experimental_p1:<30.3f}")
print("-" * 50)
```
